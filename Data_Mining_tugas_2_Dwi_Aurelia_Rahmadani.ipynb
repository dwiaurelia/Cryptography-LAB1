```python
# Import library
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('/content/iris.data', header=None,
                 names=['sepal_length','sepal_width','petal_length','petal_width','species'])
df.head()
```

```python
# Melihat informasi umum dataset
df.info()
```

```python
# Statistik deskriptif
df.describe()
```

```python
# Distribusi target (species)
df['species'].value_counts(normalize=True)
```

```python
# Analisis Univariat: distribusi tiap fitur numerik
df[['sepal_length','sepal_width','petal_length','petal_width']].hist(figsize=(10,8), bins=15)
plt.show()
```

```python
# Distribusi kategori species
sns.countplot(x='species', data=df)
plt.title("Distribusi Species")
plt.show()
```

```python
# Analisis Bivariat: scatterplot antar fitur
sns.scatterplot(x='sepal_length', y='petal_length', hue='species', data=df)
plt.title("Sepal Length vs Petal Length")
plt.show()
```

```python
# Boxplot petal_length per species
sns.boxplot(x='species', y='petal_length', data=df)
plt.title("Petal Length berdasarkan Species")
plt.show()
```

```python
# Analisis Multivariat: pairplot
sns.pairplot(df, hue='species')
plt.show()
```
